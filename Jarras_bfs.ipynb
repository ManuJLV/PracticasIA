```python
class BreadthFirstSearch:
  def __init__(self, problem):
    self.open=[]
    self.close=[]
    self.children=[]
    self.problem=problem

  def run(self):
    self.open.append(self.problem.initial)
    solution=[]
    while self.open :
      actual=self.open.pop(0)
      print('actual',actual)
      if(actual == self.problem.goal):
        while actual:
          action=actual.action
          actual=actual.parent
          solution.append(action)
        solution.reverse()
        return solution

      else:
        self.close.append(actual)
        self.children=self.problem.expand(actual)
        self.clean()
        self.open.extend(self.children)
    return solution

  def clean(self):
    for n in self.children[:]:
      for m in self.open:
        if(n==m):
          self.children.remove(n)
    for n in self.children[:]:
      for m in self.close:
        if(n==m):
          self.children.remove(n)
```

```python
class Problem(object):

  def __init__(self, initial, goal):

    if not isinstance(initial,Node):
      raise TypeError('node type is required for initial')
    if not isinstance(goal,Node):
      raise TypeError('node type is required for initial')

    self.initial = initial
    self.goal = goal

  def expand(self, node):

    raise NotImplementedError
```

```python
class Node:
    def __init__(self, state, parent=None, action=None, path_cost=0):
        self.state = state
        self.parent = parent
        self.action = action
        self.path_cost = path_cost
        self.depth = 0
        if parent:
            self.depth = parent.depth + 1

    def __repr__(self):
      rep = 'Node(' + str(self.state)+ ', ' + str(self.action) +')'
      return rep


    def __eq__(self, node):
        return self.state == node.state
```

```python
class Jarra(Problem):
  def __init__(self, initial, goal):
    Problem.__init__(self, initial, goal)


  def expand(self, node):
    children = []
    llenarA = self.la(node)
    if llenarA is not None:
      children.append(llenarA)

    llenarB=self.lb(node)
    if llenarB is not None:
      children.append(llenarB)

    vaciarA=self.va(node)
    if vaciarA is not None:
      children.append(vaciarA)

    vaciarB=self.vb(node)
    if vaciarB is not None:
      children.append(vaciarB)

    vaciarBaA=self.vba(node)
    if vaciarBaA is not None:
      children.append(vaciarBaA)

    vaciarAaB=self.vab(node)
    if vaciarAaB is not None:
      children.append(vaciarAaB)


    return children

#Llenar la jarra
  def la(self,node):
    state = node.state
    if state[0] != 5:
      newState = (5,state[1])
      return Node(newState, node,'llenar jarra 5')
    return None

  def lb(self,node):
    state = node.state
    if state[1] != 3:
      newState = (state[0], 3)
      return Node(newState,node,'llenar jarra 3')
    return None
#vaciar jarra
  def va(self,node):
    state = node.state
    if state[0] != 0:
      newState = (0,state[1])
      return Node(newState,node,'vaciar jarra 5')
    return None

  def vb(self, node):
    state = node.state
    if state[1] != 0:
      newState = (state[0],0)
      return Node(newState, node,'vaciar jarra 3')
    return None

#intercambiar jarra
  def vab(self, node):
    state = node.state
    if state[0] > 0 and state[1] <= 3:
      add = 3 - state[1]
      newState = (0, state[1] + state[0]) if add >= state[0] else  (state[0]-add, state[1]+add)
      return Node(newState, node, 'vaciar jarra-5 en jarra-3')
    return None

  def vba(self, node):
    state = node.state
    if state[1] > 0 and state[0] <= 5:
      add = 5 - state[0]
      newState = (state[0] + state[1], 0) if add >= state[1] else (state[0] + add, state[1] - add )
      return Node(newState, node, 'vaciar jarra-3 en jarra-5')
    return None
```

```python
def main():
    initial = Node([0,0])
    goal= Node([4,3])

    jarra = Jarra(initial,goal)

    bfs=BreadthFirstSearch(jarra)
    solution=bfs.run()
    print(solution)


if __name__ == "__main__":
    main()
```
