```python
#Importamos las librerias a usar
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import io
from google.colab import files
from mlxtend.frequent_patterns import association_rules
from mlxtend.frequent_patterns import fpgrowth
```

```python
uploaded = files.upload()
```

```python
# Decode the byte stream to a string
csv_data = uploaded['transacciones.csv'].decode('utf-8')  # Assuming UTF-8 encoding
# Create a StringIO object from the decoded string
csv_buffer = io.StringIO(csv_data)
# Read the CSV data using pd.read_csv
df = pd.read_csv(csv_buffer)
```

```python
df
```

```python
df = df.drop('transaccion', axis=1)
```

```python
#Transformar todos los datos a datos binomiales
df = df.applymap(lambda x: True if x > 0 else False)
```

```python
df
```

```python
#Ejecutar el algoritmo FP-Growth con sun soporte mínimo del 15%
frequent_itemsets = fpgrowth(df, min_support=0.15, use_colnames=True)
```

```python
#Imprima los itemset frecuentes
frequent_itemsets
```

```python
#Genere las reglas de asociación. Tenga en cuenta que se manejará una confianza del 70%
num_itemsets = len(frequent_itemsets)
reglas = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7, num_itemsets=len(frequent_itemsets), return_metrics=['support', 'confidence'])
```

```python
#Imprima las reglas generadas con su respectivo soporte y confianza
reglas
```
